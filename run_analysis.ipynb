## Model: nllb-200-3.3B

```python
build_dataset(attn_root="outputs/nllb-200-3.3B/attentions/", output_csv="results/3.3B/eigen_other_eng.csv", mode=2, topk=12, limit_per_pair=None, overwrite=False)
```

### Run analysis

```python
df = pd.read_csv("results/3.3B/eigen_eng_other.csv")
run_analysis(df, output_dir="results/3.3B/eng_other", n_splits=5, seeds=[42, 1, 2], name="3.3B_eng_other")
```

```python
df = pd.read_csv("results/3.3B/eigen_other_eng.csv")
run_analysis(df, output_dir="results/3.3B/other_eng", n_splits=5, seeds=[42, 1, 2], name="3.3B_other_eng")
```

### Leave-one-out

```python
df = pd.read_csv("results/3.3B/eigen_eng_other.csv")
leave_one_out(df, output_dir="results/3.3B/eng_other_loo", name="3.3B_eng_other_loo", seed=42, lang_col="tgt_lang")
```

```python
df = pd.read_csv("results/3.3B/eigen_other_eng.csv")
leave_one_out(df, output_dir="results/3.3B/other_eng_loo", name="3.3B_other_eng_loo", seed=42, lang_col="src_lang")
```
